```python
from rotpy.system import SpinSystem
from PIL import Image

# 1. Crear sistema
system = SpinSystem()

# 2. Actualizar interfaces
system.update_interface_list()
interfaces = system.create_interface_list()

n_ifaces = interfaces.get_size()
print("Interfaces detectadas:", n_ifaces)

if n_ifaces == 0:
    raise RuntimeError("No se detectaron interfaces")

# 3. Tomar la primera interfaz
iface = interfaces.create_interface(0)

# 4. Actualizar cámaras en esa interfaz
iface.update_camera_list()
cams = iface.create_camera_list()

n_cams = cams.get_size()
print("Cámaras detectadas:", n_cams)

if n_cams == 0:
    raise RuntimeError("No se detectaron cámaras")

# 5. Tomar la primera cámara
cam = cams.create_camera(0)

# 6. Inicializar y adquirir
cam.init()
cam.begin_acquisition()

frame = cam.get_next_image()

img = frame.asarray()
timestamp = frame.timestamp

print("Timestamp:", timestamp)

Image.fromarray(img).save("foto_rotpy.png")

# 7. Limpieza
frame.release()
cam.end_acquisition()
cam.deinit()
```

```python
from rotpy.system import SpinSystem
from rotpy.camera import Camera
from PIL import Image

# 1. Sistema
system = SpinSystem()
system.update_interface_list()
interfaces = system.create_interface_list()

if interfaces.get_size() == 0:
    raise RuntimeError("No se detectaron interfaces")

# 2. Interfaz
iface = interfaces.create_interface(0)
iface.update_camera_list()

# 3. Obtener TL NodeMap de la interfaz
tl_nodemap = iface.get_tl_node_map()

# 4. Obtener CameraList desde el nodemap
camera_list_node = tl_nodemap.get_node("CameraList")

n_cams = camera_list_node.get_num_children()
print("Cámaras detectadas:", n_cams)

if n_cams == 0:
    raise RuntimeError("No se detectaron cámaras")

# 5. Crear cámara a partir del índice 0
cam = Camera(camera_list_node.get_child(0))

# 6. Inicializar y adquirir
cam.init()
cam.begin_acquisition()

frame = cam.get_next_image()

img = frame.asarray()
timestamp = frame.timestamp

print("Timestamp:", timestamp)

Image.fromarray(img).save("foto_rotpy.png")

# 7. Limpieza
frame.release()
cam.end_acquisition()
cam.deinit()
iface.release()
```
